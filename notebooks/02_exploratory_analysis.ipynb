```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
```

```python
PROCESSED_DIR = Path("../data/processed")
```

```python
customers = pd.read_csv(PROCESSED_DIR / "customers.csv")
subscriptions = pd.read_csv(PROCESSED_DIR / "subscriptions.csv")
payments = pd.read_csv(PROCESSED_DIR / "payments.csv")
viewing_activity = pd.read_csv(PROCESSED_DIR / "viewing_activity.csv")
support_tickets = pd.read_csv(PROCESSED_DIR / "support_tickets.csv")
customer_feedback = pd.read_csv(PROCESSED_DIR / "customer_feedback.csv")
content = pd.read_csv(PROCESSED_DIR / "content.csv")
churn_labels = pd.read_csv(PROCESSED_DIR / "churn_labels.csv")
subscription_plans = pd.read_csv(PROCESSED_DIR / "subscription_plans.csv")
```
