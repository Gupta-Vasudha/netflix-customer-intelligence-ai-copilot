```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv("../data/raw/customers.csv")

subscription_plans = pd.read_csv("../data/raw/subscription_plans.csv")

subscriptions = pd.read_csv("../data/raw/subscriptions.csv")

content = pd.read_csv("../data/raw/content.csv")

viewing_activity = pd.read_csv("../data/raw/viewing_activity.csv")

payments = pd.read_csv("../data/raw/payments.csv")

support_tickets = pd.read_csv("../data/raw/support_tickets.csv")

customer_feedback = pd.read_csv("../data/raw/customer_feedback.csv")

churn_labels = pd.read_csv("../data/raw/churn_labels.csv")
```

```python
datasets = {
    "customers": customers,
    "subscription_plans": subscription_plans,
    "subscriptions": subscriptions,
    "content": content,
    "viewing_activity": viewing_activity,
    "payments": payments,
    "support_tickets": support_tickets,
    "customer_feedback": customer_feedback,
    "churn_labels": churn_labels
}
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-"*50)
    print(df.dtypes)
```

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-" * 50)
    missing = df.isnull().sum()

    print(missing[missing > 0].sort_values(ascending=False))
```

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print(df.duplicated().sum())
```

```python
# inspecting numerical columns

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-" * 50)

    numerical_columns = df.select_dtypes(include= np.number).columns

    if len(numerical_columns) > 0:
        display(df[numerical_columns].describe().T)
```

```python
# inspect categorical columns

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-" * 50)

    categorical_columns = df.select_dtypes(include= 'object').columns

    for column in categorical_columns:
        print(f"\n{column}")
        print(df[column].value_counts(dropna= False).head(20))
```

```python
summary = []

for name, df in datasets.items():
    summary.append({
        "dataset": name.upper(),
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": df.duplicated().sum(),
        "missing_cells": df.isnull().sum().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
print("Shape:", customers.shape)

print("\nColumns:")
print(customers.columns.tolist())

print("\nData types:")
print(customers.dtypes)

print("\nMissing values:")
print(customers.isnull().sum())

print("\nDuplicate complete rows:")
print(customers.duplicated().sum())

print("\nDuplicate customer IDs:")
print(customers["customer_id"].duplicated().sum())

print("\nUnique customer IDs:")
print(customers["customer_id"].nunique())
```

```python
duplicate_rows = customers[
    customers.duplicated(keep=False)
].sort_values("customer_id")

duplicate_rows.head(20)
```

```python
duplicate_check = (
    duplicate_rows
    .groupby("customer_id")
    .size()
    .sort_values(ascending=False)
)

duplicate_check
```
